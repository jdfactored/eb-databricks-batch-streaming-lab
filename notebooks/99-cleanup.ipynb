# 99 — Clean Up

**Run this only after your work has been graded.**

It stops any streaming queries you left running, then permanently drops your schemas,
tables and volumes — including all checkpoints and generated files. There is no undo.

```python
from helpers import utils, setup_schemas, setup_volumes, streaming_utils

streaming_utils.stop_all_streams()

cfg = utils.get_configs()
print("\nThis will permanently delete:")
for layer in utils.LAYERS:
    print(f"  {cfg['catalog']}.{cfg['base_user']}_{layer}")
for volume in utils.VOLUMES:
    print(f"  /Volumes/{cfg['catalog']}/{cfg['schema_bronze']}/{volume}")
```

## Confirm

```python
CONFIRM = False

if CONFIRM:
    setup_volumes.clean_up_volumes()
    setup_schemas.clean_up_schemas()
    print("\nCleanup complete.")
else:
    print("Nothing deleted. Set CONFIRM = True to proceed.")
```
